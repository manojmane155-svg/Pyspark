```python
#The .sort() method changes the original list.
#It does not return the sorted list — it returns None.

my_list = [2,1,0,4,3]
sorted_list = my_list.sort()
print(f'sorted list: {sorted_list}, my list: {my_list}')
```

```python
#The sorted() function creates a new list.
#The original list remains unchanged.

my_list = [2,1,0,4,3]
sorted_list1 = sorted(my_list)
print(f'sorted list_new: {sorted_list1}, my list: {my_list}')
```
