```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType

spark = SparkSession.builder.master("local[*]").appName("SimpleDF").getOrCreate()

# Step 1: Data
data = [("A",), ("B",), ("C",)]

# Step 2: Schema
schema = StructType([StructField("letter", StringType(), True)])

# Step 3: Create DataFrame
df = spark.createDataFrame(data, schema)

# Step 4: Display safely
print(df.collect())        # Works
print(df.take(3))          # Works
df.toPandas()              # Works in Jupyter
```

```python
df.toPandas()
```

```python
print(df.take(5))
```

```python
# Filter
df.filter(df["letter"] == "A").toPandas()

# Count
df.groupBy().count().collect()

# Add column
from pyspark.sql.functions import lit
df = df.withColumn("flag", lit("Y"))
df.toPandas()
```
