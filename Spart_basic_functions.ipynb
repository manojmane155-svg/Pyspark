```python
Salary = { 1: 'Dept 1',
                2: 'Dept 2',
              3: 'Dept 3',
              4: 'Dept 4',
              5: 'Dept 5',
              5: 'Dept 5'}
```

```python
# Broadcast the variable
broadcast_dept_name =spark.sparkContext.broadcast(Salary)
```

```python
# check the value of the variable
broadcast_dept_name.value
```

```python
#create UDF to return Dept name
from pyspark.sql.functions  import udf, col
@udf
def get_dept_name(Salary):
    return broadcast_dept_name.value.get(Salary)
```

```python
emp = emp_data_1.unionAll(emp_data_2)
```

```python
emp.show()
```

```python
from pyspark.sql.functions import desc, asc, col
emp_sorted = emp.orderBy(col("Salary").asc())
```

```python
emp_sorted.show()
```

```python
from pyspark.sql.functions import count
emp_count = emp_sorted.groupBy("department_id").agg(count("employee_id").alias("Total_Dept_count"))
```

```python
emp_count.show()
```

```python
from pyspark.sql.functions import sum
emp_sum = emp_sorted.groupBy("department_id").agg(sum("salary").alias("Total_Dept_Salary"))
```

```python
emp_sum.show()
```

```python
from pyspark.sql.functions import avg
emp_avg= emp_sorted.groupBy("department_id").agg(avg("salary").alias("Total_AVG_salary"))
```

```python
emp_avg.show()
```

from pyspark.sql.functions import avg
emp_avg_1=emp_sorted.groupBy("department_id").agg(avg("salary").alias("Total_avg_salary").where("Total_avg_salary > 48000")

```python
from pyspark.sql.functions import avg
emp_avg_1=emp_sorted.groupBy("department_id").agg(avg("salary").alias("avg_salary")).where("avg_salary > 48000")
```

```python
emp_avg_1.show()
```

```python
from pyspark.sql.functions import avg, round
emp_avg_round=emp_sorted.groupBy("department_id").agg(round(avg("salary"),0).alias("avg_salary")).where("avg_salary > 48000")
```

```python
emp_avg_round.show()
```

```python
from pyspark.sql.functions import avg, round, col
from pyspark.sql.types import IntegerType
emp_avg_round_1=emp_sorted.groupBy("department_id").agg(round(avg("salary"),0).cast(IntegerType()).alias("avg_salary")).where("avg_salary > 48000")
```
